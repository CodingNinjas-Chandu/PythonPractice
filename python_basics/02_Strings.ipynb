### 1. How does zero-based positive indexing work in Python strings, and how does it handle non-alphanumeric characters?

**Answer:** Python strings are contiguous sequences of Unicode code points indexed from $0$ to $N - 1$, where $N = \text{len(string)}$. Accessing `string[i]` operates in $\mathcal{O}(1)$ time. Every distinct character—including spaces, punctuation, symbols, and tabs—occupies an exact positive index offset.


````python
text = "Hello, World!"
# repr() - Returns the official text form of an object.

print(f"Char at 0: {repr(text[0])}")
print(f"Char at 5 (comma): {repr(text[5])}")
print(f"Char at 6 (space): {repr(text[6])}")
print(f"Char at 12 (symbol): {repr(text[12])}")
````

### 2. What is the execution mechanism and mathematical formula for negative indexing in Python strings?

**Answer:** Negative indexing addresses characters relative to the end of the string. The last character is located at index `-1`, down to the first character at `-len(string)`. Internally, Python converts a negative index $k$ into a positive index using $k_{\text{internal}} = N + k$, retaining $\mathcal{O}(1)$ time complexity.

````python
text = "PYTHON"
n = len(text)
negative_index = -1
print(f"Last char via negative index -1: {text[negative_index]}")
print(f"Last char via positive index ({n}(len) + (-1)): {text[n + negative_index]}")
print(f"First char via negative index (-6): {text[-6]}")
````

### 3. What exception occurs when accessing an out-of-bounds negative index, and why?

**Answer:** Accessing an index outside the valid bounds $[-N, N - 1]$ causes the Python runtime to raise an `IndexError: string index out of range`. Python does not clamp out-of-bounds indices during direct index lookups (`__getitem__`), enforcing strict spatial safety.


````python
text = "Hello"  # Length = 5; valid negative range is [-5, -1]
try:
    char = text[-6]
    print(f"Character: {char}")
except IndexError as e:
    print(f"Caught expected exception: {type(e).__name__} -> {e}")
````

### 4. How does Python enforce half-open interval semantics in basic slicing (`[start:end]`)?

**Answer:** Python string slicing adheres to half-open intervals: $[ \text{start}, \text{end} )$, meaning `start` is inclusive while `end` is exclusive. The slice extracts elements with indices satisfying $\text{start} \le i < \text{end}$. The resulting substring length is exactly $\max(0, \text{end} - \text{start})$. Time complexity is $\mathcal{O}(K)$, where $K$ is the length of the slice.


````python
text = "Hello, World!"
# Indices 1 through 4 ('e', 'l', 'l', 'o')
slice_result = text[1:5]
print(f"Extracted slice [1:5]: {repr(slice_result)}")
print(f"Length of slice: 5 - 1 == {len(slice_result)} ")
````

### 5. What are the evaluation mechanics when omitting the `end` index in a slice (`string[start:]`)?

**Answer:** When `end` is omitted (or set to `None`), Python defaults `end` to the sequence length $N$ when the step is positive. Slicing begins at index `start` (inclusive) and continues through the final character of the string.


````python
text = "Hello, World!"
sub = text[7:]
print(f"Slice from index 7 to end: {repr(sub)}")
print(f"Equivalent explicit slice: {repr(text[7:len(text)])}")
````

### 6. What are the evaluation mechanics when omitting the `start` index in a slice (`string[:end]`)?

**Answer:** When `start` is omitted (or set to `None`), Python sets the starting index to $0$ for positive step values. Slicing extracts characters from the start of the string up to, but not including, index `end`.

* when the `step` value is negative, the defaults for the start and end indices are reversed so that Python traverses the string in reverse:
  * `start` defaults to the end of the string (specifically index -1 or $N - 1$, where $N$ is the length of the string).
  * `end` defaults to before the beginning of the string (specifically $-N - 1$, allowing it to include the very first character at index 0).
  * When using a negative step, your `start` index must always be greater than (to the right of) your `end` index else you will get ""/None output
  * For example, omitting both bounds with a negative step ([::-1]) successfully reverses the entire string.
  * ```
    sub = text[::-1]
    print(f"Slice from end to start (reversed): {repr(sub)}")
    print(f"Equivalent explicit slice: {repr(text[-1:-(len(text)+1):-1])}")
    sub = text[-2:-7:-1]
    print(f"Slice from index -2 to -7 (exclusive, reversed): {repr(sub)}")
    print(f"Equivalent explicit slice (11 to 6): {repr(text[11:6:-1])}")
    sub = text[1:4:-1]
    # Return empty string because start is less than end and -ve step
    print(f"Slice from index 1 to 4 (-ve step): {repr(sub)}")
    ```



````python
text = "Hello, World!"
sub = text[:5]
print(f"Slice from start to index 5 (exclusive): {repr(sub)}")
print(f"Equivalent explicit slice: {repr(text[0:5])}")
````

### 7. How does Python slicing behave when the `end` index exceeds the string length?

**Answer:** Slicing never raises an `IndexError` when indices fall outside the string boundaries. Python clips `end` to $\min(\text{end}, N)$. Any index beyond the terminal boundary is safely bounded at the end of the string.



````python
text = "Hello"
# Valid positive range is 0-4; end=10 is well beyond len(text)
clipped_slice = text[1:10]
print(f"Input string: {text}")
print(f"Slice text[1:10]: {repr(clipped_slice)}")
````

### 8. What is the return value when the `start` index of a slice is completely out of bounds?

**Answer:** When the `start` index is greater than or equal to the string length $N$ (for positive step values), Python clamps the slice indices such that $\text{start} = \text{end} = N$, returning an empty string `""` without raising an error.


````python
text = "Hello"  # len(text) == 5
empty_slice = text[6:10]
print(f"Slice text[6:10]: {repr(empty_slice)}")
print(f"Type: {type(empty_slice)}, Length: {len(empty_slice)}")
````

### 9. Why does `string[-3:0]` evaluate to an empty string under default forward slicing?

**Answer:** Slicing with a positive step (default `step = 1`) requires the condition $\text{start} < \text{end}$ within the normalized index space. In `"Python"`, index `-3` resolves to index $3$ (`'h'`). Index `0` refers to the start of the string. Because the sequence cannot progress forward from index $3$ to index $0$, Python produces an empty string.

````python
text = "Python"
# -3 normalizes to len(text) + (-3) = 6 - 3 = 3
# Slicing from index 3 to index 0 with step +1 is contradictory
result = text[-3:0]
print(f"Slice text[-3:0]: {repr(result)}")
print(f"Normalized comparison: start=3, end=0 -> yields empty: {result == ''}")
````

### 10. How does the `step` parameter operate during positive extended slicing (`string[start:end:step]`)?

**Answer:** The `step` parameter dictates the stride index increment between sampled characters: $i_k = \text{start} + k \times \text{step}$ for $k \ge 0$ while $i_k < \text{end}$. A step of $2$ selects every second character, skipping the intermediate element.


````python
text = "Hello, World!"
# Extracts indices 1, 3, 5, 7, 9
stepped = text[1:10:2]
print(f"Slice text[1:10:2]: {repr(stepped)}")
````

### 11. How does Python interpret negative start and end bounds in slicing (`string[-k:]` and `string[:-k]`)?

**Answer:** Negative indices are normalized using $N + \text{index}$. `string[-k:]` normalizes to `string[N - k:N]`, selecting the final $k$ characters. Conversely, `string[:-k]` normalizes to `string[0:N - k]`, selecting the entire string except the final $k$ characters.


````python
text = "Python Programming is Fun!"
last_four = text[-4:]   # Last 4 characters
all_but_last_five = text[:-5]  # Drop the last 5 characters

print(f"text[-4:]: {repr(last_four)}")
print(f"text[:-5]: {repr(all_but_last_five)}")
````

### 12. How does `string[::-1]` reverse a string, and what are its algorithmic characteristics?

**Answer:** When `step` is negative (`-1`), Python inverts default boundary logic: omitted `start` defaults to $N - 1$, and omitted `end` defaults to an index representing the position before index $0$. Characters are visited in reverse order, creating a reversed copy in $\mathcal{O}(N)$ time and $\mathcal{O}(N)$ space.


````python
text = "Python Programming is Fun!"
reversed_text = text[::-1]
print(f"Original: {text}")
print(f"Reversed: {reversed_text}")
````

### 13. What is the algorithmic complexity of `len()` on a string, and what characters are counted?

````python
text = "Hi, there"
total_length = len(text)
print(f"Text: {repr(text)}")
print(f"Calculated length: {total_length}")
````

### 14. What are the operational mechanics and return type of `str.upper()`?

**Answer:** `str.upper()` returns a newly allocated string where all cased characters are converted to uppercase based on Unicode case mappings. Characters without case definitions (such as numbers and punctuation) remain unchanged. Complexity is $\mathcal{O}(N)$ time and space.


````python
text = "Hi, there"
upper_text = text.upper()
print(f"Original: {text}")
print(f"str.upper(): {upper_text}")
````

### 15. What are the operational mechanics and return type of `str.lower()`?

**Answer:** `str.lower()` returns a newly allocated string where all cased characters are converted to lowercase according to the Unicode character database. Non-cased characters are left unaltered. The operation runs in $\mathcal{O}(N)$ time and memory.

````python
text = "Hi, there"
lower_text = text.lower()
print(f"Original: {text}")
print(f"str.lower(): {lower_text}")
````

### 16. How does `str.title()` handle casing across word boundaries?

**Answer:** `str.title()` converts the first character of each word to uppercase and all remaining cased characters to lowercase. Words are identified as contiguous sequences of letters delimited by non-alphabetic characters.


````python
text = "Python Programming is Fun!"
title_text = text.title()
print(f"Original: {text}")
print(f"str.title(): {title_text}")
````

### 17. How does `str.capitalize()` differ from `str.title()`?

**Answer:** `str.capitalize()` transforms only the very first character of the entire string to uppercase, while forcing all subsequent characters to lowercase. `str.title()` capitalizes the first character of *every* word across the string.


````python
text = "Python Programming is Fun!"
capitalized = text.capitalize()
print(f"Original: {text}")
print(f"str.capitalize(): {capitalized}")
````

### 18. How does the binary `+` operator execute string concatenation?

**Answer:** When both operands are strings, the `+` operator invokes `str.__add__(self, other)`. It allocates a new memory buffer of size $L_1 + L_2$ and copies both strings into it, returning a new string in $\mathcal{O}(L_1 + L_2)$ time.


````python
str1 = "Hello"
str2 = "World"
result = str1 + str2
print(f"str1: {repr(str1)}, str2: {repr(str2)}")
print(f"Result (str1 + str2): {repr(result)}")
````

### 19. Why must numeric values be explicitly converted using `str()` prior to `+` concatenation?

**Answer:** Python is strongly typed and does not perform implicit type coercion between numeric types and strings during `+` evaluation. To concatenate, the numeric operand must be explicitly converted using `str()` to generate an intermediate string representation.

````python
name = "John"
age = 22
message = "Hello, " + name + "! You are " + str(age) + " years old."
print(f"Concatenated message: {message}")
````

### 20. What exception is raised when attempting direct addition between a string and a number?

**Answer:** Attempting to evaluate `str + int` or `str + float` raises a `TypeError`. Python's `str.__add__` checks for string instances and returns `NotImplemented`, causing the runtime to raise a `TypeError: can only concatenate str (not "int") to str`.


````python
age = 36.0
try:
    txt = "My name is Alice, I am " + age
except TypeError as e:
    print(f"Caught expected exception: {type(e).__name__} -> {e}")
````

### 21. How does the string repetition operator (`*`) function internally?

**Answer:** The `*` operator invokes `str.__mul__(self, n)`. It allocates a buffer of size $\text{len}(s) \times n$ and replicates the character sequence $n$ times. If $n \le 0$, it returns an empty string `""`. The complexity is $\mathcal{O}(N \times n)$.


````python
str1 = "abc"
result = str1 * 3
print(f"Repeated 3 times: {repr(result)}")
print(f"Repeated 0 times: {repr(str1 * 0)}")
````

### 22. What underlying values dictate character-by-character string comparison in Python?

**Answer:** Relational string comparisons evaluate characters sequentially based on their Unicode ordinal code points (`ord(char)`). Evaluation compares corresponding characters index-by-index until a mismatch occurs or one string terminates.

````python
char_a = 'a'
char_b = 'b'
# ord() converts a single character into its number(Unicode code point).
print(f"ord('a'): {ord(char_a)}")
print(f"ord('b'): {ord(char_b)}")
print(f"'apple' < 'banana': {'apple' < 'banana'}")
````

### 23. Why do uppercase ASCII characters evaluate as smaller than lowercase ASCII characters?

**Answer:** In the ASCII/Unicode standard, uppercase Latin letters `A-Z` occupy code points 65–90, whereas lowercase `a-z` occupy code points 97–122. Because comparisons are strictly ordinal, every uppercase ASCII letter is ordered before every lowercase ASCII letter.

````python
upper_z = 'Z'
lower_a = 'a'
print(f"ord('Z'): {ord(upper_z)}, ord('a'): {ord(lower_a)}")
print(f"'Z' < 'a': {upper_z < lower_a}")
print(f"'Apple' < 'apple': {'Apple' < 'apple'}")
````

### 24. How does Python evaluate relational comparisons between strings that share a common prefix?

**Answer:** If two strings share identical characters up to index $k - 1$, the comparison is determined by the first divergent character at index $k$. If one string terminates without a mismatch, the shorter string is considered smaller.

````python
s1 = "apple"
s2 = "apricot"
# Common prefix 'ap'; divergent index 2 has 'p' (ord 112) vs 'r' (ord 114)
print(f"Character comparison: {repr(s1[2])} ({ord(s1[2])}) vs {repr(s2[2])} ({ord(s2[2])})")
print(f"'apple' > 'apricot': {s1 > s2}")
````

### 25. What boolean results are produced by comparing identical vs divergent strings using relational operators?

**Answer:** Python supports six relational operators on strings: `==`, `!=`, `<`, `<=`, `>`, and `>=`. Equality checks require identical lengths and identical code points at every position.

````python
s1 = "apple"
s2 = "banana"
print(f"{s1} == {s2}: {s1 == s2}")
print(f"{s1} != {s2}: {s1 != s2}")
print(f"{s1} < {s2}:  {s1 < s2}")
print(f"{s1} > {s2}:  {s1 > s2}")
print(f"{s1} <= {s2}: {s1 <= s2}")
print(f"{s1} >= {s2}: {s1 >= s2}")
````

### 26. How does string immutability affect the behavior of mutating methods like `str.replace()`?

**Answer:** Python strings are immutable; their internal character buffers cannot be modified in place. Methods such as `replace()` do not change the calling string object; instead, they construct and return a completely new string in memory.

````python
original = "Hello, world!"
mutated_reference = original.replace("world", "Python")
print(f"Original remains intact: {repr(original)}")
print(f"New object created:     {repr(mutated_reference)}")
print(f"Are objects identical?  {original is mutated_reference}")
````

### 27. How does `str.replace(old, new)` replace multi-character substrings?

**Answer:** `str.replace(old, new)` searches for non-overlapping occurrences of the substring `old` and replaces them with `new`. If no maximum replacement count is specified, it replaces all occurrences throughout the string.

````python
sentence = "Hello, world! The world is large."
#string.replace(old, new, count(no. of times to replace - optional))
replaced = sentence.replace("world", "Python",1)
print(f"Substrings replaced: {repr(replaced)}")
````

### 28. How does `str.replace(old, new)` operate on single-character substitutions?

**Answer:** When both `old` and `new` are single characters, `str.replace()` scans the string and substitutes every matching character. It processes every instance in the string, maintaining $\mathcal{O}(N)$ linear time complexity.

````python
original = "Hello, world!"
replaced = original.replace('o', 'x')
print(f"Original: {repr(original)}")
print(f"Replaced 'o' with 'x': {repr(replaced)}")
````

### 36. What does `str.index(sub)` return when the target substring is present?

### 37. What exception is raised when `str.index(sub)` fails to locate the substring?

### 38. How does `str.find(sub)` report the location of a found substring?

### 47. How does `str.isupper()` evaluate strings with non-cased characters?

**Answer:** `str.isupper()` returns `True` if all cased characters in the string are uppercase and there is at least one cased character. Non-cased characters (digits, symbols, whitespace) are ignored during the check.


````python
print(f"'HELLO'.isupper():     {'HELLO'.isupper()}")
print(f"'Hello'.isupper():     {'Hello'.isupper()}")
print(f"'HELLO 123!'.isupper(): {'HELLO 123!'.isupper()}")  # True: digits/symbols ignored
print(f"'123!'.isupper():       {'123!'.isupper()}")        # False: no cased characters
````

````python
print(f"'hello'.islower():     {'hello'.islower()}")
print(f"'Hello'.islower():     {'Hello'.islower()}")
print(f"'hello 123!'.islower(): {'hello 123!'.islower()}")  # True: ignores non-cased characters
print(f"'123!'.islower():       {'123!'.islower()}")        # False: no cased characters
````

### 49. What characters satisfy `str.isnumeric()`, and how does it treat whitespace?

**Answer:** `str.isnumeric()` returns `True` if all characters in the string are numeric and there is at least one character. This includes standard digits as well as Unicode numeric characters (such as fractions and superscripts). Whitespace characters cause `isnumeric()` to return `False`.

````python
t1 = "12345"
t2 = "123abc"
t3 = "123 456"

print(f"{repr(t1)}.isnumeric(): {t1.isnumeric()}")
print(f"{repr(t2)}.isnumeric(): {t2.isnumeric()}")
print(f"{repr(t3)}.isnumeric(): {t3.isnumeric()}")
````

### 50. How does `str.isdigit()` differ from `str.isnumeric()` on digits and whitespace?

**Answer:** `str.isdigit()` returns `True` if all characters are digits (characters with the Unicode numeric property value `Digit`) and there is at least one character. Whitespace and letters return `False`. While all digits are numeric, some numeric characters (like certain fractions) are not digits.

````python
s1 = "12345"
s2 = "abc123"
s3 = "123 456"

print(f"{repr(s1)}.isdigit(): {s1.isdigit()}")
print(f"{repr(s2)}.isdigit(): {s2.isdigit()}")
print(f"{repr(s3)}.isdigit(): {s3.isdigit()}")
````

### 51. How does `str.swapcase()` transform cased vs non-alphanumeric characters?

**Answer:** `str.swapcase()` returns a newly allocated string where all lowercase characters are converted to uppercase and all uppercase characters are converted to lowercase. Non-cased characters (digits, spaces, punctuation) remain untouched.

````python
text = "Hello World! 123"
swapped = text.swapcase()
print(f"Original: {repr(text)}")
print(f"Swapped:  {repr(swapped)}")
````
