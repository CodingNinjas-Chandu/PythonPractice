### 1. How does the `print()` function work, and how can it print multiple objects?

**Answer:** The `print()` function displays output to the screen and accepts one or more comma-separated objects enclosed in parentheses. When multiple comma-separated items are passed, it automatically inserts a space between them.

```python
# Printing single and multiple objects
print("Hello, World!")
print("Hi", "Hello")
```

### 2. How do you modify line endings in `print()` using the `end` parameter?

**Answer:** By default, `print()` appends a newline character (`\n`) after executing. The `end` parameter allows you to specify a custom string (such as a space, hyphen, or empty string) to print at the end instead of creating a new line.


```python
# Replacing default newline with custom characters
print("Hello", end=" ")
print("World")

print("Python", end="-")
print("programming")
```

### 3. How do you customize item separators in `print()` using the `sep` parameter?

**Answer:** The `sep` parameter specifies the separator character placed between multiple comma-separated arguments passed to the `print()` function, overriding the default single space.

```python
# Customizing the separator
print("apple", "banana", "orange", sep=", ")
print("2026", "10", "12", sep="-")
```

### 6. Why is indentation important in Python, and how does it differ from languages like Java or C++?

**Answer:** Unlike C++ or Java which define blocks using curly braces (`{}`), Python uses indentation (consistent whitespace) to define the scope and structure of code blocks like conditionals and loops.


```python
x = 5
if x > 3:
    # Indented block belongs to the if statement
    print("x is greater than 3")
```

### 7. What are variables, and why are they fundamental in programming?

**Answer:** Variables are named memory containers used for storing data. They allow storing data for later use, manipulating values, reusing information, and improving readability.



```python
# Variables store and reuse data
radius = 5
pi = 3.14159
area = pi * (radius ** 2)
print("Calculated Area:", area)
```

### 8. How are variables declared and initialized in Python?

**Answer:** Python does not use explicit variable declaration keywords (like `let` or `int`). A variable is automatically created and initialized the moment a value is assigned to it using the `=` operator.

```python
# Automatic declaration upon assignment
x = 42
my_name = "John"
print(x, my_name)
```

### 9. What are the rules for naming variable identifiers in Python?

**Answer:** Identifiers can contain letters (`a-z`, `A-Z`), digits (`0-9`), and underscores (`_`), but cannot begin with a digit. They cannot use spaces or special characters (like hyphens), are case-sensitive, and cannot match reserved Python keywords.

```python
# Legal variable names
my_variable = 42
this_is_a_variable = "Hello"
number_1 = 10

# Illegal examples:
# 1_number = 42      (starts with digit)
# my-variable = 10   (contains hyphen)
# for = 10           (reserved keyword)
print(my_variable, this_is_a_variable, number_1)
```

### 10. How does case sensitivity apply to variable names?

**Answer:** Python identifiers are case-sensitive, meaning variable names differing only in capitalization refer to distinct variables in memory.


```python
Temp = 98.6
temp = 37.0

print(f"Temp: {Temp}, temp: {temp}")
print("Temp is temp:", Temp is temp)
```

### 11. What happens when a variable is reassigned to a new value?

**Answer:** Variables can vary their contents. When a new value is assigned, the variable points to the new value, replacing the old reference.



```python
Var = 10
print("Initial:", Var)

Var = "Hello"
print("Reassigned:", Var)
```

### 12. How do you perform multiple variable assignment (unpacking) in a single line?

**Answer:** Python allows assigning comma-separated values to multiple comma-separated variables in a single line.

### 13. How do you assign a single value to multiple variables simultaneously?

**Answer:** Multiple variables can be chained together with assignment operators (`=`) to assign them the same value in one line.

### 14. What is the difference between variables, identifiers, and keywords?

**Answer:**

* **Identifiers**: General names given to programming elements like variables, functions, and classes.

* **Variables**: A specific type of identifier used to store data values in memory.

* **Keywords**: Reserved words with predefined language meanings (e.g., `if`, `for`, `def`, `class`) that cannot be used as identifiers.


```python
import keyword
# List all reserved keywords
print(keyword.kwlist)
print(f"\nTotal keywords: {len(keyword.kwlist)}")

# Identifier naming a variable
user_count = 10

# Checking reserved status of words
print("'def' is keyword:", keyword.iskeyword("def"))
print("'print' is keyword:", keyword.iskeyword("print"))
print("'True' is keyword:", keyword.iskeyword("True"))
print("'assert' is keyword:", keyword.iskeyword("assert"))
print("'user_count' is keyword:", keyword.iskeyword("user_count"))
```

### 15. What are the primary numeric data types in Python?

**Answer:** The primary numeric types are `int` (whole numbers), `float` (numbers with decimal points), and `complex` (numbers with real and imaginary parts represented with `j`).

```python
x = 10       # int
y = 3.14     # float
z = 2 + 3j   # complex

print(type(x), type(y), type(z))
print(f"Complex real: {z.real}, imag: {z.imag}")
```

### 16. What are strings, and what does immutability mean for them?

**Answer:** Strings (`str`) are sequences of characters enclosed within single or double quotes. They are immutable, meaning individual characters cannot be modified in-place after creation.

```python
name = "John"

# Accessing characters is allowed:
print("First char:", name[0])

# Attempting in-place modification raises a TypeError:
try:
    name[0] = "R"
except TypeError as e:
    print(f"Caught error: {e}")
```

### 17. How do you resolve quote conflicts within string literals?

**Answer:** Quote conflicts can be resolved by using double quotes around single quotes, escaping quotes with a backslash (`\`), or enclosing the text in triple quotes (`"""`).

```python
# 1. Opposite quotes
print("She is Ana's sister")

# 2. Backslash escape
print('She is Ana\'s sister')
print("She said: \"Ana helps me with homework.\"")

# 3. Triple quotes
print("""She said, "Ana's homework is done".""")
```

### 18. What are the common escape characters `\n` and `\t`, and what do they do?

**Answer:** `\n` inserts a newline break, while `\t` inserts a horizontal tab to move the cursor to the next tab stop for neat columnar formatting.

```python
# Newline and tab escape characters
print("Line 1\nLine 2")
print("Name:\t\tJohn")
print("Age:\t\t25")
print("Occupation:\tEngineer")
```

### 19. What are Boolean values, and what are their valid states?

**Answer:** The `bool` data type represents truth values and can only take one of two predefined keyword states: `True` or `False`.


```python
is_active = True
is_expired = False

print(type(is_active))
print(is_active and not is_expired)
```

### 20. How does truthiness work with the `bool()` function on empty vs non-empty values?

**Answer:** The `bool()` function evaluates any object: empty values (like empty strings `""`, `0`, or `None`) evaluate to `False`, while non-empty values (like `"Hello"` or non-zero numbers) evaluate to `True`.


```python
# Truthiness evaluations
print(bool(""))       # False
print(bool(0))        # False
print(bool(None))     # False
print(bool("Hello"))  # True
print(bool(12))       # True
```

### 21. How do you inspect the data type of a variable or expression at runtime?

**Answer:** The built-in `type()` function inspects and returns the class/type of any given value or variable.

### 22. What is type conversion (typecasting)?

**Answer:** Type conversion (typecasting) is the process of converting a value explicitly from one data type to another using built-in functions like `int()`, `float()`, `str()`, and `bool()`.

```python
z="5"
print(z, type(z))

x = int(z)
print(x, type(x))

y = float(x)
print(y, type(y))  # 5.0 <class 'float'>

num_str = str(x)
print(num_str, type(num_str))  # '5' <class 'str'>
```

### 23. Why is explicit type casting necessary when capturing user numeric input?

**Answer:** Because `input()` always returns text data (`str`), any arithmetic operations require converting the input to `int` or `float` using explicit type conversion.


```python
# Simulating keyboard string input converted to int
age_str = input("Enter your age: ")         # from input("Enter your age: ")
print("age type age_str:", type(age_str))
age_int = int(age_str)

print("Next year age:", age_int + 1)
print("age type age_int:", type(age_int))
```

### 24. How does type conversion behave between Booleans and Integers?

**Answer:** Booleans subclass integers in Python; converting `True` using `int(True)` yields `1`, while `int(False)` yields `0`.

```python
flag_true = True
flag_false = False

print(int(flag_true))   # 1
print(int(flag_false))  # 0
print(bool(1))          # True
print(bool(0))          # False
```
